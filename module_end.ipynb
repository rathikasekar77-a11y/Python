 Customer Support Ticket Analyser

```python

tickets = {
    "Ticket_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Customer_Name": ["Arun", "Priya", "Rahul", "Sneha", "Karthik",
                      "Divya", "Manoj", "Anitha", "Vijay", "Meena"],
    "Issue": ["Login Problem", "Payment Failed", "Slow Internet",
              "App Crash", "Password Reset", "Refund Request",
              "Login Problem", "Network Issue", "Payment Failed",
              "App Crash"],
    "Priority": ["High", "Medium", "Low", "High", "Low",
                 "Medium", "High", "Low", "Medium", "High"],
    "Status": ["Open", "Closed", "Open", "Closed", "Open",
               "Closed", "Open", "Open", "Closed", "Open"],
    "Sentiment": ["Negative", "Neutral", "Positive", "Negative",
                  "Neutral", "Positive", "Negative", "Neutral",
                  "Negative", "Positive"]
}
print("CUSTOMER SUPPORT TICKET DETAILS")
print("--------------------------------")

for i in range(len(tickets["Ticket_ID"])):
    print("Ticket ID:", tickets["Ticket_ID"][i])
    print("Customer Name:", tickets["Customer_Name"][i])
    print("Issue:", tickets["Issue"][i])
    print("Priority:", tickets["Priority"][i])
    print("Status:", tickets["Status"][i])
    print("Sentiment:", tickets["Sentiment"][i])
    print("--------------------------------")
```

```python
n = int(input("How many new tickets do you want to add? "))
for i in range(n):

    print("\nEnter details for new ticket")

    # Auto-increment ticket number starting from 11
    ticket_id = len(tickets["Ticket_ID"]) + 101

    customer_name = input("Enter Customer Name: ")
    issue = input("Enter Issue Description: ")
while True:
        priority = input("Enter Priority (High / Medium / Low): ").capitalize()

        if priority in ["High", "Medium", "Low"]:
            break
        else:
            print("Invalid priority! Please enter High, Medium, or Low.")
tickets["Ticket_ID"].append(ticket_id)
tickets["Customer_Name"].append(customer_name)
tickets["Issue"].append(issue)
tickets["Priority"].append(priority)
tickets["Status"].append("Open")
tickets["Sentiment"].append("Neutral")

print("Ticket added successfully!")
print("\nUPDATED TICKET DETAILS")
print("----------------------------")

for i in range(len(tickets["Ticket_ID"])):
    print("Ticket ID:", tickets["Ticket_ID"][i])
    print("Customer Name:", tickets["Customer_Name"][i])
    print("Issue:", tickets["Issue"][i])
    print("Priority:", tickets["Priority"][i])
    print("Status:", tickets["Status"][i])
    print("Sentiment:", tickets["Sentiment"][i])
    print("----------------------------")

```

```python
def clean_issue(issue):
  issue = issue.lower()
  issue = issue.replace(".", "")
  issue = issue.replace(",", "")
  issue = issue.replace("!", "")
  issue = issue.replace("?", "")
  issue = issue.replace("-", "")
  issue = issue.replace("ok", "okay")
  issue = issue.strip()
  issue = " ".join(issue.split())
  return issue
for i in range(len(tickets["Issue"])):
    tickets["Issue"][i] = clean_issue(tickets["Issue"][i])
print("\nCLEANED TICKET ISSUES")
print("----------------------------")

for i in range(len(tickets["Ticket_ID"])):
    print("Ticket ID:", tickets["Ticket_ID"][i])
    print("Issue:", tickets["Issue"][i])
    print("----------------------------")
```

```python
def count_tickets_with_word(word):
   count = 0
   word = word.lower()
   for issue in tickets["Issue"]:
     if word in issue.lower().split():
       count += 1
   return count

print("\nKEYWORD-BASED ISSUE INSIGHTS")
print("--------------------------------")

print("Tickets containing 'poor':", count_tickets_with_word("poor"))

print("Tickets containing 'good':", count_tickets_with_word("good"))

print("Tickets containing 'slow':", count_tickets_with_word("slow"))

print("Tickets containing 'excellent':", count_tickets_with_word("excellent"))
```

```python
print("\nFINAL CLEANED TICKET DATA")
print("--------------------------------")

for key, value in tickets.items():
    print(key, ":", value)
```

```python
high_count = 0
medium_count = 0
low_count = 0

for priority in tickets["Priority"]:

    if priority == "High":
        high_count += 1

    elif priority == "Medium":
        medium_count += 1

    elif priority == "Low":
        low_count += 1

print("\nPRIORITY ANALYSIS")
print("--------------------------------")

print("High Priority Tickets:", high_count)
print("Medium Priority Tickets:", medium_count)
print("Low Priority Tickets:", low_count)
```

```python

max_word_count = 0
longest_ticket_index = 0

for i in range(len(tickets["Issue"])):

    word_count = len(tickets["Issue"][i].split())

    if word_count > max_word_count:
        max_word_count = word_count
        longest_ticket_index = i

print("\nLONGEST ISSUE DESCRIPTION")
print("--------------------------------")

print("Ticket Number:", tickets["Ticket_ID"][longest_ticket_index])
print("Customer Name:", tickets["Customer_Name"][longest_ticket_index])
print("Cleaned Issue:", tickets["Issue"][longest_ticket_index])
print("Word Count:", max_word_count)
```

```python
unique_words = set()
for issue in tickets["Issue"]:

    words = issue.split()

    for word in words:
        unique_words.add(word)

sorted_words = sorted(unique_words)

print("\nUNIQUE WORDS ANALYSIS")
print("--------------------------------")

print("Total Unique Words:", len(unique_words))

print("Unique Words (Sorted):")

for word in sorted_words:
    print(word)
```
